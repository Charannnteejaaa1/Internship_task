## Q1. Find the Shortest Word in a Sentence

**1. Observe:** Input is a sentence; output is the shortest word.

**2. Understand:** Split the sentence into words and choose the word with the fewest characters.

**3. Break:** 1. Split on whitespace. 2. If there are no words, return None. 3. Select the shortest word.

**4. Logic:** Keep the first word with the minimum length when there is a tie.

**5. Code:** The implementation below solves the problem without hardcoding the expected output.

**6. Test:** Run the given example and an additional test case. Expected results for these tests are documented below the code.

**Reasoning / comment:** min(..., key=len) compares word lengths and preserves the first word in a tie. The second print checks a different sentence.

**Expected test output**
```text
easy
one
```

````python
def shortest_word(sentence):
    words = sentence.split()
    if not words:
        return None
    return min(words, key=len)

sentence = "Python makes coding easy"
print(shortest_word(sentence))
print(shortest_word("one tiny elephant"))
````

## Q2. Find the Largest Number in a List

**1. Observe:** Input is a list of numbers; output is the largest number. Do not use max().

**2. Understand:** Track the largest value seen while scanning the list.

**3. Break:** 1. Handle an empty list. 2. Start with the first element. 3. Compare each remaining value and update the largest.

**4. Logic:** Return None for an empty list; otherwise update a candidate whenever a larger value appears.

**5. Code:** The implementation below solves the problem without hardcoding the expected output.

**6. Test:** Run the given example and an additional test case. Expected results for these tests are documented below the code.

**Reasoning / comment:** The variable largest stores the best value seen so far. Starting from the first item also handles lists containing only negative numbers.

**Expected test output**
```text

```

````python
def largest_number(numbers):
    if not numbers:
        return None
    largest = numbers[0]
    for number in numbers[1:]:
        if number > largest:
            largest = number
    return largest

print(largest_number([12, 45, 7, 89, 23]))
print(largest_number([-8, -3, -12]))
````

## Q7. Count Consonants in a String

**1. Observe:** Count alphabetic consonants; ignore spaces, digits, and punctuation.

**2. Understand:** A consonant is a letter from a to z that is not a vowel.

**3. Break:** 1. Lowercase the input. 2. Keep only alphabetic characters. 3. Count letters outside the vowel set.

**4. Logic:** Use isalpha() to exclude digits and symbols, then check that the character is not a vowel.

## Q8. Find the Most Frequent Character

**1. Observe:** Find the most common character, ignore spaces, and break ties by first appearance.

**2. Understand:** Count non-space characters while preserving the order in which distinct characters first occur.

**3. Break:** 1. Remove spaces. 2. Handle an empty result. 3. Count characters. 4. Select the first character with the highest count.

**4. Logic:** A dictionary records counts in insertion order; max over its keys preserves the first key in a tie.

**5. Code:** The implementation below solves the problem without hardcoding the expected output.

**6. Test:** Run the given example and an additional test case. Expected results for these tests are documented below the code.

**Reasoning / comment:** The counts dictionary keeps first-seen character order. max(..., key=counts.get) selects the highest count and leaves a tie with the earliest key.

**Expected test output**
```text
a
a
```

````python
def most_frequent_character(text):
    counts = {}
    for character in text:
        if character == " ":
            continue
        counts[character] = counts.get(character, 0) + 1
    if not counts:
        return None
    return max(counts, key=counts.get)

print(most_frequent_character("banana"))
print(most_frequent_character("a b b a"))
````

## Q9. Find the Least Frequent Character

**1. Observe:** Find the least common character, ignore spaces, and break ties by first appearance.

````python
def least_frequent_character(text):
    counts = {}
    for character in text:
        if character == " ":
            continue
        counts[character] = counts.get(character, 0) + 1
    if not counts:
        return None
    return min(counts, key=counts.get)

print(least_frequent_character("success"))
print(least_frequent_character("aabcc"))
````

## Q10. Find All Characters Appearing More Than Once

**1. Observe:** Return each repeated non-space character once, ordered by first appearance.

**2. Understand:** A character qualifies if its total count is greater than one.

**3. Break:** 1. Remove spaces. 2. Count characters. 3. Walk the original character order. 4. Add each repeated character once.

**4. Logic:** Use a count dictionary and a seen set to avoid duplicate output entries.

**5. Code:** The implementation below solves the problem without hardcoding the expected output.

**6. Test:** Run the given example and an additional test case. Expected results for these tests are documented below the code.

**Reasoning / comment:** The second pass preserves first-appearance order. The seen set ensures a repeated character is added only once.

**Expected test output**
```text
['r', 'g', 'm']
['a', 'b']
```

````python
def repeated_characters(text):
    characters = [character for character in text if character != " "]
    counts = {}
    for character in characters:
        counts[character] = counts.get(character, 0) + 1
    result = []
    seen = set()
    for character in characters:
        if counts[character] > 1 and character not in seen:
            result.append(character)
            seen.add(character)
    return result

print(repeated_characters("programming"))
print(repeated_characters("a a b b c"))
````

## Q17. Find All Words Appearing More Than Once

**1. Observe:** Return repeated words once each, ordered by first appearance, ignoring case.

**2. Understand:** Normalize words, count them, then collect repeated words in their original normalized order.

**3. Break:** 1. Split and lowercase. 2. Count each word. 3. Walk the sequence and add repeated words once.

**4. Logic:** Use a count dictionary and a seen set to preserve order and prevent duplicates.

**5. Code:** The implementation below solves the problem without hardcoding the expected output.

**6. Test:** Run the given example and an additional test case. Expected results for these tests are documented below the code.

**Expected test output**
```text
['data', 'science']

```

````python
def repeated_words(sentence):
    words = sentence.lower().split()
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    result = []
    seen = set()
    for word in words:
        if counts[word] > 1 and word not in seen:
            result.append(word)
            seen.add(word)
    return result

print(repeated_words("data science uses data and science tools"))
print(repeated_words("Go go STOP stop now"))
````

## Q18. Find the First Word That Appears Only Once

**1. Observe:** Find the earliest word with frequency one, ignoring case.

**2. Understand:** Count normalized words, then scan them from left to right for the first count of one.

**3. Break:** 1. Lowercase and split. 2. Count words. 3. Scan in order. 4. Return the first unique word or None.

**4. Logic:** The first pass builds counts; the second pass preserves sentence order.

**5. Code:** The implementation below solves the problem without hardcoding the expected output.

**6. Test:** Run the given example and an additional test case. Expected results for these tests are documented below the code.

**Reasoning / comment:** The scan returns immediately when it finds a word with count one. If no word is unique, the function returns None.

**Expected test output**
```text
mango
None
```

````python
def first_unique_word(sentence):
    words = sentence.lower().split()
    counts = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    for word in words:
        if counts[word] == 1:
            return word
    return None

print(first_unique_word("apple banana apple mango banana orange"))
print(first_unique_word("yes YES"))
````

## Q20. Find the Longest Word That Appears Only Once

**1. Observe:** Find the longest word occurring exactly once, ignoring case; choose the first on a length tie.

**2. Understand:** Count lowercase forms but preserve the original order and spelling for the returned word.

**3. Break:** 1. Split into original words. 2. Count lowercase forms. 3. Keep words with count one. 4. Select the longest, preserving first on ties.

**4. Logic:** Use normalized counts for uniqueness, then max over qualifying original words using their lengths.

**5. Code:** The implementation below solves the problem without hardcoding the expected output.

**6. Test:** Run the given example and an additional test case. Expected results for these tests are documented below the code.

**Reasoning / comment:** Counts are case-insensitive, but the original word sequence is retained. max with key=len returns the first word when lengths tie.

**Expected test output**
```text
coding
cat
```

````python
def longest_unique_word(sentence):
    original_words = sentence.split()
    counts = {}
    for word in original_words:
        normalized = word.lower()
        counts[normalized] = counts.get(normalized, 0) + 1
    unique_words = [word for word in original_words
                    if counts[word.lower()] == 1]
    if not unique_words:
        return None
    return max(unique_words, key=len)

print(longest_unique_word("Python makes coding useful Python"))
print(longest_unique_word("cat dog fox"))
````
